```python
import pandas as pd
import requests



airlines_url = "https://en.wikipedia.org/wiki/List_of_airline_codes"
airport_url = "https://en.wikipedia.org/wiki/List_of_international_airports_by_country"



headers = {"User-Agent": "airline-codes-scraper/0.1 (your@email.com)"}
resp = requests.get(airlines_url, headers=headers, timeout=30)
resp.raise_for_status()
```

```python
from io import StringIO
import pandas as pd

tables = pd.read_html(StringIO(resp.text))
print(len(tables))
```

```python
df_airlines = tables[0]

print(df_airlines.head())
```
